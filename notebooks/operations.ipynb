##### Operations:
1. **Total delivered orders**
2. **Order to be delivered**
3. **Average delivery time**


```python
import polars as pl

orders = pl.read_csv(source = "../data/orders.csv",
                    schema_overrides = {"order_purchase_timestamp": pl.Datetime,
                                        "order_approved_at": pl.Datetime,
                                        "order_delivered_carrier_date": pl.Datetime,
                                        "order_delivered_customer_date": pl.Datetime,
                                        "order_estimated_delivery_date": pl.Datetime})
orders
```

```python
delivered_orders = (orders.filter(pl.col("order_status")=="delivered")
                          .get_column("order_status")      
                          .count())
delivered_orders
```

```python
delivery_time = orders.filter(pl.col("order_status")=="delivered").with_columns(
    (pl.col("order_delivered_customer_date")-pl.col("order_approved_at")).dt.total_hours().alias("dur")
).get_column("dur").mean()
delivery_time
```

```python
round(delivery_time/24)
```
